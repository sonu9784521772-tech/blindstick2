````
# Step 2: Export YOLOv8n to TFLite
from ultralytics import YOLO
import os

print('Loading YOLOv8n model...')
model = YOLO('yolov8n.pt')  # auto-downloads from ultralytics

print('Exporting to TFLite...')
result = model.export(format='tflite', imgsz=640)
print(f'Export result: {result}')
````

````
# Step 3: Find and download the .tflite file
import os
from google.colab import files

tflite_path = None
for root, dirs, filenames in os.walk('.'):
    for f in filenames:
        if f.endswith('.tflite'):
            candidate = os.path.join(root, f)
            if 'float32' in f or tflite_path is None:
                tflite_path = candidate

if tflite_path:
    size_mb = os.path.getsize(tflite_path) / (1024*1024)
    print(f'✅ Found: {tflite_path} ({size_mb:.1f} MB)')
    # Copy to root with simple name
    import shutil
    shutil.copy2(tflite_path, 'yolov8n.tflite')
    print('⬇️  Downloading yolov8n.tflite ...')
    files.download('yolov8n.tflite')
    print('✅ Download started! Save this file.')
else:
    print('❌ No .tflite file found. Check export output above.')
````
